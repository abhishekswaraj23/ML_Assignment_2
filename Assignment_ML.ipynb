```python
import pandas as pd

# Load dataset
df = pd.read_csv("data/test_data.csv")

# 1. Dataset shape
print("Dataset Shape:", df.shape)

# 2. First 5 rows
print("\nFirst 5 rows:")
display(df.head())

# 3. Column names
print("\nColumns:")
print(df.columns.tolist())

# 4. Data types
print("\nData Types:")
print(df.dtypes)

# 5. Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 6. Duplicate rows
print("\nDuplicate Rows:", df.duplicated().sum())

# 7. Target distribution
print("\nTarget Distribution:")
print(df["Diagnosis"].value_counts())

# 8. Target percentage
print("\nTarget Percentage:")
print(df["Diagnosis"].value_counts(normalize=True) * 100)

# 9. Basic statistics
print("\nStatistical Summary:")
display(df.describe())
```

```python
from sklearn.model_selection import train_test_split

# Copy dataset
data = df.copy()

# Remove ID because it is only an identifier
data = data.drop(columns=["ID"])

# Encode target
# B = 0 (Benign)
# M = 1 (Malignant)
data["Diagnosis"] = data["Diagnosis"].map({"B": 0, "M": 1})

# Separate features and target
X = data.drop(columns=["Diagnosis"])
y = data["Diagnosis"]

# Stratified 80/20 train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())

print("\nFeature count:", X_train.shape[1])
```

```python
from sklearn.preprocessing import StandardScaler

# Create scaler
scaler = StandardScaler()

# Fit ONLY on training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform test data using the same scaler
X_test_scaled = scaler.transform(X_test)

print("Original training shape:", X_train.shape)
print("Scaled training shape:", X_train_scaled.shape)

print("Original testing shape:", X_test.shape)
print("Scaled testing shape:", X_test_scaled.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# Create Logistic Regression model
logistic_model = LogisticRegression(max_iter=1000, random_state=42)

# Train the model
logistic_model.fit(X_train_scaled, y_train)

# Predictions
y_pred_lr = logistic_model.predict(X_test_scaled)

# Probability for AUC
y_prob_lr = logistic_model.predict_proba(X_test_scaled)[:, 1]

print("Logistic Regression trained successfully.")
```

```python
from sklearn.metrics import (
    accuracy_score,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    matthews_corrcoef
)

# Calculate metrics
lr_accuracy = accuracy_score(y_test, y_pred_lr)
lr_auc = roc_auc_score(y_test, y_prob_lr)
lr_precision = precision_score(y_test, y_pred_lr)
lr_recall = recall_score(y_test, y_pred_lr)
lr_f1 = f1_score(y_test, y_pred_lr)
lr_mcc = matthews_corrcoef(y_test, y_pred_lr)

print("Logistic Regression Results")
print("---------------------------")
print("Accuracy :", round(lr_accuracy, 4))
print("AUC      :", round(lr_auc, 4))
print("Precision:", round(lr_precision, 4))
print("Recall   :", round(lr_recall, 4))
print("F1 Score :", round(lr_f1, 4))
print("MCC      :", round(lr_mcc, 4))
```

```python
from sklearn.tree import DecisionTreeClassifier

# Create Decision Tree model
decision_tree_model = DecisionTreeClassifier(
    random_state=42
)

# Train
decision_tree_model.fit(X_train, y_train)

# Predictions
y_pred_dt = decision_tree_model.predict(X_test)

# Probability for AUC
y_prob_dt = decision_tree_model.predict_proba(X_test)[:, 1]

print("Decision Tree trained successfully.")
```

```python
# Decision Tree metrics

dt_accuracy = accuracy_score(y_test, y_pred_dt)
dt_auc = roc_auc_score(y_test, y_prob_dt)
dt_precision = precision_score(y_test, y_pred_dt)
dt_recall = recall_score(y_test, y_pred_dt)
dt_f1 = f1_score(y_test, y_pred_dt)
dt_mcc = matthews_corrcoef(y_test, y_pred_dt)

print("Decision Tree Results")
print("---------------------")
print("Accuracy :", round(dt_accuracy, 4))
print("AUC      :", round(dt_auc, 4))
print("Precision:", round(dt_precision, 4))
print("Recall   :", round(dt_recall, 4))
print("F1 Score :", round(dt_f1, 4))
print("MCC      :", round(dt_mcc, 4))
```

```python
from sklearn.neighbors import KNeighborsClassifier

# Create KNN model
knn_model = KNeighborsClassifier(n_neighbors=5)

# Train
knn_model.fit(X_train_scaled, y_train)

# Predictions
y_pred_knn = knn_model.predict(X_test_scaled)

# Probability for AUC
y_prob_knn = knn_model.predict_proba(X_test_scaled)[:, 1]

print("KNN trained successfully.")
```

```python
# KNN metrics

knn_accuracy = accuracy_score(y_test, y_pred_knn)
knn_auc = roc_auc_score(y_test, y_prob_knn)
knn_precision = precision_score(y_test, y_pred_knn)
knn_recall = recall_score(y_test, y_pred_knn)
knn_f1 = f1_score(y_test, y_pred_knn)
knn_mcc = matthews_corrcoef(y_test, y_pred_knn)

print("KNN Results")
print("-----------")
print("Accuracy :", round(knn_accuracy, 4))
print("AUC      :", round(knn_auc, 4))
print("Precision:", round(knn_precision, 4))
print("Recall   :", round(knn_recall, 4))
print("F1 Score :", round(knn_f1, 4))
print("MCC      :", round(knn_mcc, 4))
```

```python
from sklearn.naive_bayes import GaussianNB

# Create Gaussian Naive Bayes model
naive_bayes_model = GaussianNB()

# Train
naive_bayes_model.fit(X_train, y_train)

# Predictions
y_pred_nb = naive_bayes_model.predict(X_test)

# Probability for AUC
y_prob_nb = naive_bayes_model.predict_proba(X_test)[:, 1]

print("Naive Bayes trained successfully.")
```

```python
# Naive Bayes metrics

nb_accuracy = accuracy_score(y_test, y_pred_nb)
nb_auc = roc_auc_score(y_test, y_prob_nb)
nb_precision = precision_score(y_test, y_pred_nb)
nb_recall = recall_score(y_test, y_pred_nb)
nb_f1 = f1_score(y_test, y_pred_nb)
nb_mcc = matthews_corrcoef(y_test, y_pred_nb)

print("Naive Bayes Results")
print("-------------------")
print("Accuracy :", round(nb_accuracy, 4))
print("AUC      :", round(nb_auc, 4))
print("Precision:", round(nb_precision, 4))
print("Recall   :", round(nb_recall, 4))
print("F1 Score :", round(nb_f1, 4))
print("MCC      :", round(nb_mcc, 4))
```

```python
from sklearn.svm import SVC

# Create SVM model
svm_model = SVC(
    probability=True,
    random_state=42
)

# Train
svm_model.fit(X_train_scaled, y_train)

# Predictions
y_pred_svm = svm_model.predict(X_test_scaled)

# Probability for AUC
y_prob_svm = svm_model.predict_proba(X_test_scaled)[:, 1]

print("SVM trained successfully.")
```

```python
# SVM metrics

svm_accuracy = accuracy_score(y_test, y_pred_svm)
svm_auc = roc_auc_score(y_test, y_prob_svm)
svm_precision = precision_score(y_test, y_pred_svm)
svm_recall = recall_score(y_test, y_pred_svm)
svm_f1 = f1_score(y_test, y_pred_svm)
svm_mcc = matthews_corrcoef(y_test, y_pred_svm)

print("SVM Results")
print("-----------")
print("Accuracy :", round(svm_accuracy, 4))
print("AUC      :", round(svm_auc, 4))
print("Precision:", round(svm_precision, 4))
print("Recall   :", round(svm_recall, 4))
print("F1 Score :", round(svm_f1, 4))
print("MCC      :", round(svm_mcc, 4))
```

```python
# Final model comparison

results = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Decision Tree",
        "KNN",
        "Naive Bayes",
        "SVM"
    ],
    "Accuracy": [
        lr_accuracy,
        dt_accuracy,
        knn_accuracy,
        nb_accuracy,
        svm_accuracy
    ],
    "AUC": [
        lr_auc,
        dt_auc,
        knn_auc,
        nb_auc,
        svm_auc
    ],
    "Precision": [
        lr_precision,
        dt_precision,
        knn_precision,
        nb_precision,
        svm_precision
    ],
    "Recall": [
        lr_recall,
        dt_recall,
        knn_recall,
        nb_recall,
        svm_recall
    ],
    "F1 Score": [
        lr_f1,
        dt_f1,
        knn_f1,
        nb_f1,
        svm_f1
    ],
    "MCC": [
        lr_mcc,
        dt_mcc,
        knn_mcc,
        nb_mcc,
        svm_mcc
    ]
})

results.round(4)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay

models_predictions = {
    "Logistic Regression": y_pred_lr,
    "Decision Tree": y_pred_dt,
    "KNN": y_pred_knn,
    "Naive Bayes": y_pred_nb,
    "SVM": y_pred_svm
}

fig, axes = plt.subplots(1, 5, figsize=(20, 4))

for ax, (name, predictions) in zip(axes, models_predictions.items()):
    ConfusionMatrixDisplay.from_predictions(
        y_test,
        predictions,
        display_labels=["Benign", "Malignant"],
        ax=ax
    )
    ax.set_title(name)

plt.tight_layout()
plt.show()
```

```python
import joblib
import os

# Create model folder
os.makedirs("model", exist_ok=True)

# Save trained models
joblib.dump(logistic_model, "model/logistic_regression.joblib")
joblib.dump(decision_tree_model, "model/decision_tree.joblib")
joblib.dump(knn_model, "model/knn.joblib")
joblib.dump(naive_bayes_model, "model/naive_bayes.joblib")
joblib.dump(svm_model, "model/svm.joblib")

# Save scaler used by scaled models
joblib.dump(scaler, "model/scaler.joblib")

print("All models and scaler saved successfully.")
```
